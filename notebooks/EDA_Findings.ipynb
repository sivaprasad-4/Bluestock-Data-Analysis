# EDA Findings – Mutual Fund Analysis

# 10 Key Findings

1. **SIP inflows showed strong long-term growth:**  
   Monthly SIP inflows increased substantially from approximately ₹11,500 Cr in early 2022 to **₹31,002 Cr by December 2025**, indicating strong and sustained growth in SIP participation.

2. **Average NAV increased steadily over time:**  
   The average daily NAV across the analyzed schemes showed an overall upward trend from approximately **₹207 in January 2022 to nearly ₹358 by mid-2026**.

3. **Liquid funds dominated category-wise net inflows:**  
   The Liquid category recorded the highest net inflows throughout the analyzed period, significantly exceeding other categories.

4. **Flexi Cap and Sectoral/Thematic funds also attracted significant inflows:**  
   Flexi Cap and Sectoral/Thematic categories showed considerably higher net inflows compared with smaller categories such as ELSS and Gilt.

5. **Investors aged 26–35 formed the largest investor group:**  
   The 26–35 age group accounted for approximately **41.1%** of the investor records, making it the dominant demographic.

6. **T30 cities contributed the majority of SIP investors:**  
   Approximately **66.6%** of SIP investor records belonged to T30 cities, while B30 cities accounted for approximately **33.4%**.

7. **SIP investment was geographically distributed across 12 states:**  
   Madhya Pradesh recorded the highest aggregate SIP amount in the analyzed data, followed by Punjab and Telangana.

8. **Total folio count more than doubled during the analysis period:**  
   Total folios increased from **13.26 Cr in January 2022 to 26.12 Cr in December 2025**, indicating substantial expansion in the investor base.

9. **NAV return correlations among the 10 selected funds were generally weak:**  
   Pairwise correlations were mostly close to zero, with the strongest observed relationships still relatively low. This indicates limited linear correlation among the selected funds' daily NAV returns in the analyzed dataset.

10. **Banking, IT and Pharma were the largest sectors by combined portfolio weight:**  
    Across the analyzed equity fund holdings, Banking had the highest combined sector weight, followed by IT and Pharma. These values represent aggregated weights across funds rather than the allocation of a single portfolio.

## Overall Conclusion

The EDA indicates strong growth in mutual fund participation and SIP adoption between 2022 and 2025. SIP inflows and folio counts increased substantially, while investors aged 26–35 represented the largest share of the investor base. T30 cities contributed the majority of SIP investors. Liquid funds dominated category-wise inflows, while Banking, IT and Pharma were among the largest sectors in the aggregated equity holdings. The low return correlations among the selected schemes also indicate that their daily NAV returns did not exhibit strong linear relationships during the analyzed period.
